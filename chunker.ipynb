```python
# Fixed chunker
chunk_list = []

def chunker(df):
    max_chars=15000
    context=10
    start_size=50
    total_rows = len(df)
    end = 0
    chunk_count = 0

    while end < total_rows:
        stop = min(end + start_size, total_rows)
        if end == 0:
            df_chunk = df.iloc[end:stop]
            ctx = 0
        else:
           
            df_chunk = df.iloc[max(end - context, 0):stop]
            ctx = context
        while len(df_chunk.to_string(index=False)) > max_chars and len(df_chunk) > start_size:
            df_chunk = df_chunk.iloc[:-1]  # remove last row
            stop -= 1  

        chunk_list.append(df_chunk)
        chunk_count += 1

        num_new_rows = len(df_chunk) - ctx
        if num_new_rows <= 0:
            num_new_rows = 1
        end += num_new_rows

        print(f"Chunk {chunk_count}: rows {df_chunk.index[0]} to {df_chunk.index[-1]} ({len(df_chunk)} rows)")

    print(f"Total chunks created: {chunk_count}")
```

```python
chunker(df)
print("done")
```

```python
# check
all_rows = pd.concat(chunk_list, ignore_index=True)

# Count unique rows by RecordNumber
unique_rows = all_rows['RecordNumber'].nunique()
test1 = df['RecordNumber'].nunique()
print(f"Unique rows: {test1}")  # should match original 9389
print(f"Unique rows: {unique_rows}")  # should match original 9389
```
