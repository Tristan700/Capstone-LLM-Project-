```python
# checks to view all columns in the data set
import csv
from Evtx.Evtx import Evtx
import xml.etree.ElementTree as ET
def Convert_evtx11_to_csv(file_name):
    write = file_name.split(".")[0]
    write = os.path.join("Suspicious_logs", write+ ".csv")
    read = os.path.join("Suspicious_logs",file_name)
    print(read)


        # XML namespace
    ns = {"e": "http://schemas.microsoft.com/win/2004/08/events/event"}
    
    with Evtx(read) as log:
        # Collects field_names
        field_names = set(["RecordNumber", "Timestamp", "EventID"])
    
        for record in log.records():
            xml_data = record.xml()
            root = ET.fromstring(xml_data)
            for data in root.findall(".//e:Data", ns):
                field_names.add(data.attrib.get("Name", "Unknown"))
    
        field_names = list(field_names)  # convert to list for writing CSV
    
        #writes CSV file
        with open(write, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=field_names)
            writer.writeheader()
    # get all the data and writes it 
            with Evtx(read) as log2:
                for record in log2.records():
                    xml_data = record.xml()
                    root = ET.fromstring(xml_data)
                    
                    event = {
                        "RecordNumber": record.record_num(),
                        "Timestamp": root.find(".//e:TimeCreated", ns).attrib.get("SystemTime", ""),
                        "EventID": root.find(".//e:EventID", ns).text if root.find(".//e:EventID", ns) is not None else ""
                    }
    
                    # Add all EventData fields
                    for data in root.findall(".//e:Data", ns):
                        field = data.attrib.get("Name", "Unknown")
                        value = data.text if data.text else ""
                        event[field] = value
    
                    writer.writerow(event)
    
    print( write)
```

```python
import os
import csv
from Evtx.Evtx import Evtx
import xml.etree.ElementTree as ET

def Tconvert_evtx_to_csv(file_name):
    write_name = os.path.splitext(file_name)[0] + ".csv"
    write_path = os.path.join("Logs_Normal", write_name)
    read_path = os.path.join("Logs_Normal", file_name)
    print(f"Reading EVTX file: {read_path}")

    ns = {"e": "http://schemas.microsoft.com/win/2004/08/events/event"}

    # Columns to keep
    keep_columns = [
        "RecordNumber","Timestamp","EventID","EventName","MessageText",
        "ProcessId","ProcessCreationTime","ParentProcessName","CommandLine",
        "User","TargetUserName","SubjectUserName","IpAddress","WorkstationName",
        "Operation","Reason","ErrorCode","Component","Url","CallerProcessName",
        "TargetLogonGuid","ForceLogoff","LockoutObservationWindow","MemberName",
        "AuthenticationPackageName","ProcessName","ModuleName","ModulePath",
        "ModuleVersion","AppName","AppPath","AppVersion","Event","StorePath",
        "PackageFullName","PackageFamily"
    ]

    with Evtx(read_path) as log:
        with open(write_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=keep_columns)
            writer.writeheader()

            for record in log.records():
                xml_data = record.xml()
                root = ET.fromstring(xml_data)

                # Base event info
                event = {col: "" for col in keep_columns}
                event["RecordNumber"] = record.record_num()
                event["Timestamp"] = root.find(".//e:TimeCreated", ns).attrib.get("SystemTime", "")
                event["EventID"] = root.find(".//e:EventID", ns).text if root.find(".//e:EventID", ns) is not None else ""
                provider = root.find(".//e:Provider", ns)
                event["EventName"] = provider.attrib.get("Name", "") if provider is not None else ""
                message = root.find(".//e:RenderingInfo/e:Message", ns)
                event["MessageText"] = message.text if message is not None else ""

                # Fill all <Data Name="..."> values if in keep_columns
                for data in root.findall(".//e:Data", ns):
                    field_name = data.attrib.get("Name", "")
                    value = data.text if data.text else ""
                    if field_name in keep_columns:
                        event[field_name] = value

                writer.writerow(event)

    print(f"CSV successfully written to: {write_path}")
```
